```python
'''3.Write a Python function to check if a given string is a palindrome.
Expected Output (for input "madam"):
madam is a palindrome'''
word =input("Enter the word")
revWord=word[::-1]
if (word==revWord):
    print(f"{word} is a palindrom") 
else:
    print(f"{word} is not a palindrom") 
```

```python
'''4.Write a Python program to count the number of vowels in a given string. 
Expected Output (for input "education"):

Number of vowels: 5'''
word=input("Enter the word")
vowels=word.count('a')+word.count('e')+word.count('i')+word.count('o')+word.count('u')+word.count('A')+word.count('E')+word.count('I')+word.count('O')+word.count('U')
                                                                
print("Number of vowels",vowels) 
```

```python
'''5.Write a Python program to remove duplicates from a list.
Expected Output (for input [1, 2, 2, 3, 4, 4, 5]):

 [1, 2, 3, 4, 5]'''
listNumber=[int(x) for x in input("Enter the list values separated by space").split()]
print("list is",listNumber)
remDuplicate=list(set(listNumber))
print("After removing duplicates",remDuplicate)


```
