```python
 #3.Find Even and Odd Numbers in a List
#Question: Write a Python program to separate even and odd numbers from the list [12, 17, 19, 24, 35, 40, 56].
genList=[12, 17, 19, 24, 35, 40, 56]
evenList=[]
oddList=[]
for i in genList:
    if i%2==0:
        evenList.append(i)
    else:
        oddList.append(i)
print("Original List",genList)
print("Even Numbers",evenList)
print("Odd Numbers",oddList)
```

```python
#4. Reverse a Number
#Question: Write a Python program to reverse the number 12345.
num=input("Enter a number")
print(num)
revNum=num[::-1]
print(int(revNum))
```
