```python
class OddEven:
    def oddeven(num):
        if(num%2==0):
            return "Even"
        else:
            return "Odd"    
```

```python
n=int(input("Enter the number"))
print(OddEven.oddeven(n))
```

```python
# Create a function that tells elegibility of marriage for male and female according to their age limit like 21 for male and 18 for female
class ElegiblityForMarriage:
    def eligibility(gender,age):
        if(gender=="M" and age>=21):
            return("Eligible")
        elif(gender=="F" and age>=18):
            return("Eligible")
        else:
            return("Not Eligible")    
```

```python
g=input("Enter your Gender(M/F)")
a=int(input("Enter your age"))
print("Your Gender:",g)
print("Your Age:",a)
print("Eligibility for Marriage:",ElegiblityForMarriage.eligibility(g,a))
```

```python
# calculate the percentage of your 10th mark
class FindPercent:
    def calcPercentage(listMarks):
        sum=0
        for i,marks in enumerate(listMarks):
            print("Subject",i+1,":",marks)
            sum=sum+marks
        print("Total:",sum)
        return((sum/500)*100)
      
```

```python
Sub1=int(input("Enter your sub1 Mark"))
Sub2=int(input("Enter your sub2 Mark"))
Sub3=int(input("Enter your sub3 Mark"))
Sub4=int(input("Enter your sub4 Mark"))
Sub5=int(input("Enter your sub5 Mark"))
listMarks=[Sub1,Sub2,Sub3,Sub4,Sub5]
print("Percentage:",FindPercent.calcPercentage(listMarks))
```

```python
#print area and perimeter of triangle using functions
class triangle:
    def triangle():
        height=float(input("Enter the height of the triangle"))
        breath=float(input("Enter the breath of the triangle"))
        print("Area formula: (Height*Breadth)/2")
        print("Area of the triangle:",(height*breath)/2)
        side1=float(input("Enter the side1 length of the triangle"))
        side2=float(input("Enter the side2 length of the triangle"))
        print("Perimeter formula: side1+side2+breath")
        print("Perimeter of the triangle",side1+side2+breath)
```
