```python
#1.System checks if user is eligible to vote based on their age. Write a logic to ask the user their age and determine 
#if they are eligible to vote based on whether they are 18 or not
age=int(input("Enter your age"))
if (age<18):
    print("Not Eligible for voting")
else:
    print("Eligible for voting")
```

```python
#2.A Program processes a list of numbers and need to find the largest value
listNum=[77,22,3,34,67,23,90,76,45,65,39,76,40,21,22]
highestNum=0
for i in listNum:
    if i>highestNum:
        highestNum=i
print("Highest number among the given list is",highestNum)
```

```python
#3.A company provides employees with a 10%bonus if their salary exceeds $50000
sal=float(input("Enter your salary"))
if sal>50000:
    print("Congratulations! You receive 10% bonus as well")
    sal=sal+(sal*0.1)
    print("Revised salary is",sal)
else:
    print("You will be awarded in the coming cycles")
        
```

```python
#4.A program evaluates a number to determine if it is even or odd. Write logic to tcheck if number is even or odd
eveOddNum=int(input("Enter the number"))
if eveOddNum%2==0:
    print("The given number",eveOddNum,"is Even")
else:
    print("The given number",eveOddNum,"is Odd")
```

```python
#5 Text processing tool reverses a given word or sentence for formatting purpose
#Write a logic to take word or sentence as input and produce its reversed version
word=input("Enter a word")
revWord=""
revWord=word[::-1]
print("Word:",word)
print("Reverse of the Word:",revWord)
```

```python
#6. Grading system determines whether student has passed or failed based on their scores. Write logic to check 
#if student has passed a subject by scoring at least 40 marks
marks=float(input("Enter the marks"))
if marks>=40:
    print("Congratulations!, You have passed the subject")
else:
    print("Failed! Try again!")
```

```python
#7. Retail store offers a 20% discount if a customer's total order exceeds $100. Write logic to calcualte the final amount
#to be paid after applying the discount
totalAmount=float(input("Enter the total order amount"))
if totalAmount>=100:
    totalAmount=totalAmount-(totalAmount*0.2)
    print("Congratulations! you have got 20% discount on your purchase!")
    print("Your New Total is",totalAmount)
```

```python
#8. A banking system processes withdrawal requests and ensures the user has enough balance 
currBal=float(input("Enter your current Balance"))
withDrawAmount=float(input("Enter the amount you want to withdraw"))
remAmount=currBal-withDrawAmount
if remAmount <= 1000:
    print("Your new balnace",remAmount," is less than minimum balance limit! Cannot withdraw")
else:
    print("Withdrawal success")
```
